```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


file_path = '/content/yt_data_sb.csv'
data = pd.read_csv(file_path)

data_cleaned = data.drop(columns=["Unnamed: 0"])

essential_columns = ["video id", "title", "trending date", "publish time"]
data_cleaned = data_cleaned.dropna(subset=essential_columns)

numeric_columns = ["views", "likes", "dislikes", "comment count"]
data_cleaned[numeric_columns] = data_cleaned[numeric_columns].fillna(0)

data_cleaned["trending date"] = pd.to_datetime(data_cleaned["trending date"], format="%y.%d.%m", errors='coerce')

data_cleaned["publish time"] = pd.to_datetime(data_cleaned["publish time"], errors='coerce')

data_cleaned = data_cleaned[(data_cleaned['views'] >= 0) &
                            (data_cleaned['likes'] >= 0) &
                            (data_cleaned['dislikes'] >= 0)]

plt.figure(figsize=(10, 6))
plt.hist(data_cleaned["views"], bins=50, color='skyblue', edgecolor='black')
plt.title("Distribution of Video Views")
plt.xlabel("Views")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


plt.figure(figsize=(10, 6))
plt.hist(data_cleaned["likes"], bins=50, color='lightgreen', edgecolor='black')
plt.title("Distribution of Video Likes")
plt.xlabel("Likes")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


plt.figure(figsize=(10, 6))
plt.hist(data_cleaned["dislikes"], bins=50, color='salmon', edgecolor='black')
plt.title("Distribution of Video Dislikes")
plt.xlabel("Dislikes")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


data_cleaned.set_index('publish time', inplace=True)
average_views_by_date = data_cleaned['views'].resample('M').mean()

plt.figure(figsize=(12, 6))
plt.plot(average_views_by_date, color='blue')
plt.title("Trend of Average Views Over Time")
plt.xlabel("Publish Time")
plt.ylabel("Average Views")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


data_cleaned.reset_index(inplace=True)


top_channels = data_cleaned.groupby("channel title")["views"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
top_channels.plot(kind='bar', color='purple', edgecolor='black')
plt.title("Top 10 Channels by Total Views")
plt.xlabel("Channel Title")
plt.ylabel("Total Views")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


correlation_data = data_cleaned[["views", "likes", "dislikes", "comment count"]]
correlation_matrix = correlation_data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix between Views, Likes, Dislikes, and Comment Count")
plt.show()


weekly_views = data_cleaned.set_index("publish time")["views"].resample("W").sum()

plt.figure(figsize=(12, 6))
plt.plot(weekly_views, color='teal')
plt.title("Weekly Trend in Video Views")
plt.xlabel("Publish Time (Weekly)")
plt.ylabel("Total Views")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


top_categories = data_cleaned.groupby("category id")["views"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
top_categories.plot(kind='bar', color='orange', edgecolor='black')
plt.title("Top 10 Video Categories by Total Views")
plt.xlabel("Category ID")
plt.ylabel("Total Views")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


data_cleaned['publish_hour'] = data_cleaned['publish time'].dt.hour
data_cleaned['publish_day'] = data_cleaned['publish time'].dt.day_name()
publish_time_heatmap = pd.crosstab(data_cleaned['publish_day'], data_cleaned['publish_hour'])

plt.figure(figsize=(12, 6))
sns.heatmap(publish_time_heatmap, cmap="YlGnBu", linewidths=0.5)
plt.title("Video Publish Time Distribution (Hour vs. Day)")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")
plt.show()

plt.figure(figsize=(12, 6))
sns.scatterplot(data=data_cleaned, x="publish_hour", y="views", alpha=0.5, color="orange")
plt.title("Scatter Plot of Views vs. Publish Hour")
plt.xlabel("Publish Hour")
plt.ylabel("Views")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


data_cleaned['likes_dislikes_ratio'] = data_cleaned['likes'] / (data_cleaned['dislikes'] + 1)
plt.figure(figsize=(12, 6))
sns.boxplot(data=data_cleaned, x="category id", y="likes_dislikes_ratio", palette="viridis")
plt.title("Likes-to-Dislikes Ratio by Category")
plt.xlabel("Category ID")
plt.ylabel("Likes-to-Dislikes Ratio")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
